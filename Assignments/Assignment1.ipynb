```python
import pandas as pd
data = pd.read_csv('/content/drive/MyDrive/MSSP6070/Assignment1/data_academic_performance.csv')
```

New groups of parents' education level

```python
education_map = {
    'Postgraduate education': 'Graduate Degree',
    'Complete professional education': 'Bachelor’s Degree',
    'Incomplete Professional Education': 'Some College or Associate Degree',
    'Complete technique or technology': 'Some College or Associate Degree',
    'Complete Secundary': 'High School Completion',
    'Incomplete technical or technological': 'Less than High School',
    'Incomplete Secundary': 'Less than High School',
    'Complete primary': 'Less than High School',
    'Incomplete primary': 'Less than High School',
    'Ninguno': 'No Formal Education'
}
data['Father_EducationGroup'] = data['EDU_FATHER'].map(education_map)
data['Mother_EducationGroup'] = data['EDU_MOTHER'].map(education_map)
data[['EDU_FATHER', 'Father_EducationGroup', 'EDU_MOTHER', 'Mother_EducationGroup']].head()
```

Mean quantitative reasoning score based on fathers' education level

```python
mean_quant_by_father_edu = data.groupby('Father_EducationGroup')['QR_PRO'].mean().round(2).sort_values(ascending=False)
mean_quant_by_father_edu
```

Mean quantitative reasoning score based on mothers' education level

```python
mean_quant_by_mother_edu = data.groupby('Mother_EducationGroup')['QR_PRO'].mean().round(2).sort_values(ascending=False)
mean_quant_by_mother_edu
```

Mean critical reading score based on fathers' education level

```python
mean_cr_by_father_edu = data.groupby('Father_EducationGroup')['CR_PRO'].mean().round(2).sort_values(ascending=False)
mean_cr_by_father_edu
```

Mean critical reading score based on mothers' education level

```python
mean_cr_by_mother_edu = data.groupby('Mother_EducationGroup')['CR_PRO'].mean().round(2).sort_values(ascending=False)
mean_cr_by_mother_edu
```

Outcomes based on Stratum

```python
mean_scores_by_Stratum = (
    data[data['STRATUM'] != '0']
    .groupby('STRATUM')[['QR_PRO', 'CR_PRO', 'CC_PRO', 'ENG_PRO', 'WC_PRO', 'FEP_PRO']]
    .mean()
    .round(2)
)
mean_scores_by_Stratum
```

Outcomes based on Socioeconomic Levels

```python
mean_scores_by_SEL = (
    data.groupby('SEL')[['QR_PRO', 'CR_PRO', 'CC_PRO', 'ENG_PRO', 'WC_PRO', 'FEP_PRO']]
    .mean()
    .round(2)
)
mean_scores_by_SEL
```

Mean Global Score Based on Gender

```python
mean_scores_by_Gender = (
    data.groupby('GENDER')[['G_SC']]
    .mean()
    .round(2)
)
mean_scores_by_Gender
```

Median Global Score Based on Gender

```python
median_scores_by_Gender = (
    data.groupby('GENDER')[['G_SC']]
    .median()
    .round(2)
)
median_scores_by_Gender
```
